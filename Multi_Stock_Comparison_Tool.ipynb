# 📈 Multi-Stock Comparison Tool
### Real-Time Financial Analysis and Portfolio Visualization

Compare relative performance, daily returns, and moving averages across multiple stocks using `yfinance` and interactive visualizations.

---

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from datetime import datetime, timedelta

# Note: yfinance would be used for real data fetching
# !pip install yfinance

import plotly.io as pio
pio.templates.default = "plotly_dark"
```

## 1. Mock Data Source
Simulating stock price movements for demonstration.

```python
def get_stock_data(tickers, days=180):
    end_date = datetime.now()
    start_date = end_date - timedelta(days=days)
    dates = pd.date_range(start=start_date, end=end_date, freq='B')
    
    all_data = []
    for ticker in tickers:
        # Random walk for prices
        start_price = np.random.uniform(50, 500)
        returns = np.random.normal(0.0005, 0.012, len(dates))
        prices = start_price * (1 + returns).cumprod()
        
        df = pd.DataFrame({'Date': dates, 'Ticker': ticker, 'Price': prices})
        all_data.append(df)
    
    return pd.concat(all_data)

tickers = ['AAPL', 'GOOGL', 'TSLA', 'MSFT', 'AMZN']
df = get_stock_data(tickers)
df.head()
```

## 2. Normalized Performance Comparison
Visualizing percentage returns from a common starting point.

```python
df['Normalized_Return'] = df.groupby('Ticker')['Price'].transform(lambda x: (x / x.iloc[0]) - 1)

fig = px.line(df, x='Date', y='Normalized_Return', color='Ticker', 
              title='Relative Stock Growth (Cumulative Return)',
              labels={'Normalized_Return': 'Cumulative Return (%)'})
fig.update_layout(yaxis_tickformat='.1%')
fig.show()
```

## 3. Correlation Heatmap
Understanding asset relationships.

```python
pivot_df = df.pivot(index='Date', columns='Ticker', values='Price')
corr = pivot_df.pct_change().corr()

fig = px.imshow(corr, text_auto=True, title='Daily Returns Correlation Matrix', 
                color_continuous_scale='RdBu_r', range_color=[-1, 1])
fig.show()
```
